# Functions with outputs : keyword is "return"

```python
def format_name():
    first_name = input("Enter your first name: ")
    last_name = input("Enter your last name: ")
    return first_name.title() + " " + last_name.title()

full_name = format_name()
print("Formatted Name:", full_name)
```

```python
def function_1(text):
    return text + text

def function_2(text):
    return text.title()

output = function_2(function_1("heLLo")) # Calling one function inside another function
print(output)  # Output: Hellohello
```

```python
# Check Leap year
def is_leap_year(year):
    if year%4 == 0:
        if year%100 == 0:
            if year%400 == 0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False

print(is_leap_year(2100))
print(is_leap_year(1600))
print(is_leap_year(2032))
```

```python
# Docstrings
def is_leap_year(year):
    """This function checks if a given year is a leap year or not.
    A leap year is divisible by 4, but not divisible by 100 unless it is also divisible by 400."""

    if year%4 == 0:
        if year%100 == 0:
            if year%400 == 0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False

print(is_leap_year.__doc__)
print(is_leap_year(2100))
```

```python
# let's build calculator with functions and outputs
def add(n1, n2):
    return n1 + n2

def subtract(n1, n2):
    return n1 - n2

def multiply(n1, n2):
    return n1 * n2

def divide(n1, n2):
    return n1 / n2

operations = {
    "+": add,
    "-": subtract,
    "*": multiply,
    "/": divide 
}

def calculator():
    print("Welcome to the calculator!")
    calc = "y"
    num1 = float(input("Enter first number: "))
    while calc == "y":        
        for symbol in operations:
            print(symbol)

        operation_symbol = input("Pick an operation from the line above: ")
        num2 = float(input("Enter second number: "))
        answer = operations[operation_symbol](num1, num2)
        print(f"{num1} {operation_symbol} {num2} = {answer}")

        calc = input("Type 'y' to continue calculating with the answer, or type 'n' to start a new calculation: ")
        if calc == "y":
            num1 = answer
        else:
            calc = "n"
    print('Final answer of the calculation is:', answer)
    print("Thank you for using the calculator!")

calculator()
```
