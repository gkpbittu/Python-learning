# Randomisation and Python Lists

```python
import random
```

```python
random_integer = random.randint(1, 10)
random_float = random.random()
random_float_range = random.uniform(1, 5)
random_choice = random.choice(['apple', 'banana', 'cherry'])
random_shuffle = random.shuffle([1, 2, 3, 4, 5])
random_sample = random.sample(range(1, 100), 5)
random_seed = random.seed(42)  # Set seed for reproducibility
random_gauss = random.gauss(0, 1)  # Gaussian distribution with mean=0 and stddev=1
random_expovariate = random.expovariate(1/5)  # Exponential distribution with lambda=1/5
random_betavariate = random.betavariate(2, 5)  # Beta distribution with alpha=2 and beta=5
random_weibullvariate = random.weibullvariate(1, 1.5)  # Weibull distribution with alpha=1 and beta=1.5
random_paretovariate = random.paretovariate(1.5)  # Pareto distribution with alpha=1.5
random_vonmisesvariate = random.vonmisesvariate(0, 1)  # Von Mises distribution with mu=0 and kappa=1
random_gammavariate = random.gammavariate(2, 2)  # Gamma distribution with alpha=2 and beta=2
random_lognormvariate = random.lognormvariate(0, 1)  # Log-normal distribution with mu=0 and sigma=1
random_triangular = random.triangular(1, 10, 5)  # Triangular distribution with low=1, high=10, mode=5
random_choices = random.choices(['red', 'green', 'blue'], weights=[0.1, 0.3, 0.6], k=5)  # Weighted random choices
random_sample_with_replacement = random.choices(range(1, 100), k=5)  # Sample with replacement
random_sample_without_replacement = random.sample(range(1, 100), 5)  # Sample without replacement
```

```python
print("Random Integer:", random_integer)
print("Random Float:", random_float)
print("Random Float in Range:", random_float_range)
print("Random Choice:", random_choice)
print("Random Shuffle:", random_shuffle)
print("Random Sample:", random_sample)
print("Random Seed:", random_seed)
print("Random Gauss:", random_gauss)
print("Random Exponential Variate:", random_expovariate)
print("Random Beta Variate:", random_betavariate)
print("Random Weibull Variate:", random_weibullvariate)
print("Random Pareto Variate:", random_paretovariate)
print("Random Von Mises Variate:", random_vonmisesvariate)
print("Random Gamma Variate:", random_gammavariate)
print("Random Log-normal Variate:", random_lognormvariate)
print("Random Triangular Variate:", random_triangular)
print("Random Choices with Weights:", random_choices)
print("Random Sample with Replacement:", random_sample_with_replacement)
print("Random Sample without Replacement:", random_sample_without_replacement)
```

# Toss game: Heads/Tails

```python
# Approach1
random_heads_or_tails = random.randint(a=0, b=1)
if random_heads_or_tails == 0:
    result = "heads"
else:
    result = "tails"
print(f"The coin landed on: {result}")
```

```python
# Approach2: Generate the randome heads/tails game
print("Welcome to the Heads or Tails Game!")
choice = input("Choose heads or tails: ").lower()
if choice not in ['heads', 'tails']:
    print("Invalid choice. Please choose 'heads' or 'tails'.")
else:
    flip = random.choice(['heads', 'tails'])
    print(f"The coin landed on: {flip}")
    if choice == flip:
        print("You win!")
    else:
        print("You lose!")
```

Who will pay the bill?

```python
friends = ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Heidi", "Ivan", "Judy"]
print(friends)

random_index = random.randint(0, len(friends) - 1)
pay_due_on = friends[random_index]  # Randomly select a friend to pay due
print(f"{pay_due_on} has to pay the bill today!")
```

```python
# other way using the built-in methods of random module
pay_due_on = random.choice(friends)  # Randomly select a friend to pay due
print(f"{pay_due_on} has to pay the bill today!")
```

# Project: Rock, Paper, scissor

```python
print("Let's play a game of Rock, Paper, Scissors!")
user_choice = input("Enter your choice (rock, paper, scissors): ").lower()
system_choice = random.choice(['rock', 'paper', 'scissors'])

print(f"You chose {user_choice}.")
print(f"The system chose {system_choice}.")

if user_choice == system_choice:
    print("It's a tie!")
else:
    if (user_choice == "rock" and system_choice == "scissors") or \
       (user_choice == "paper" and system_choice == "rock") or \
       (user_choice == "scissors" and system_choice == "paper"):
        print("You win!")
    else:
        print("System wins!")
```
