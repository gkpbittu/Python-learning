```python
# Check odd or even

num = int(input('Enter the number to check odd or even'))

if num > 0:
    if num%2 != 0:
        print(f'The entered number {num} is an odd number.')
    else:
        print(f'The entered number {num} is an even number.')
else:
    print('Please enter valid positive real number to check odd or even.')
```

```python
weight = 85
height = 1.85

bmi = weight / (height ** 2)

if bmi < 18.5:
    print("underweight")
elif bmi >= 18.5 and bmi < 25:
    print("normal weight")
else:
    print('overweight')
```

```python
print('Welcome to the rollercoaster')
height = int(input('What is your height in CM?'))
bill = 0
if height >= 120:
    print(f'Your height is {height} CM, You can ride!!')
    age = int(input("What's your age?"))
    if age < 12:
        bill = 5
        print(f"Your age is {age}, child tickets are $5")
    elif age <= 18:
        bill = 7
        print(f"Your age is {age}, youth tickets are $7")
    else:
        bill = 12
        print(f"Your age is {age}, adult tickets are $12")

    wants_photo = input("Do you want to have a photo? Type y for Yes and n for No.")
    if wants_photo == 'y':
        bill += 3
        print(f"Your total bill is ${bill}")

    print(f"Your total bill is ${bill}")
        
else:
    print(f"Your height is {height} CM, sorry!! you have to grow taller before you can ride.")
```

```python
print("Welcome to the python Pizza order deliveries!")
size = input('What size pizza do you want? S, M Or L: ')
pepperoni = input("Do you want pepporoni on your pizza? Y or N: ")
extra_cheese = input("Do you want extra cheese? Y or N: ")

bill = 0
if size == 'S':
    bill = 15
    if pepperoni == 'Y':
        bill += 2
    if extra_cheese == 'Y':
        bill += 1
    print(f"you've ordered small size pizza, your bill is ${bill}")
elif size == 'M':
    bill = 20
    if pepperoni == 'Y':
        bill += 3
    if extra_cheese == 'Y':
        bill += 1
    print(f"you've ordered medium size pizza, your bill is ${bill}")
else:
    bill = 25
    if pepperoni == 'Y':
        bill += 3
    if extra_cheese == 'Y':
        bill += 1
    print(f"you've ordered large size pizza, your bill is ${bill}")
```

# Treasure Island Project

```python
print("Welcome to the treasure Island. Your mission is to find the treasure.")
step1 = input("Which direction you want to go: Left or Right?")
if step1 == 'Right':
    print("Game Over")
else:
    step2 = input("Do you want to swim or wait: Swim or Wait?")
    if step2 == 'Swim':
        print('Game Over')
    else:
        step3 = input("Which door you will choose? Red, Blue or Yellow")
        if step3 == 'Yellow':
            print("You win! Treasure Found")
        else:
            print("Game Over")
```
