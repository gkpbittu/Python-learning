# BMI Calculator

```python
height = 1.65
weight = 84

# BMI is equal to the person's weight divided by the person's height squared
BMI = weight/(height**2)
print(BMI)
```

# Project Building

```python
print("Welcome to the tip calculator")
t_bill = float(input('What was the total bill?'))
tip = float(input('How much tip would you like to give in percentage? 10,12, or 15?'))
persons_to_split = int(input('How many people to split the total bill?'))

share = (t_bill * (1 + tip/100))/persons_to_split
print(f' Each person should pay ${round(share,2)}')
```
