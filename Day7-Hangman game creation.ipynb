# Building the Hangman game

```python
import random
```

```python
wordlist = ['python', 'java', 'kotlin', 'javascript']
# 1. Randomly choose a word from the word_list
chosen_word = random.choice(wordlist)

placeholder = "-" * len(chosen_word)
print(f"Guess the word: {placeholder}")

# 2. Ask the user to guess a letter and assign their answer to a variable called guess. Make
# sure to call the lower() method on their answer.  
# enable the user to guess a letter again and again until they run out of lives or they find the word.

# Set the number of lives to the length of a random word from the wordlist
no_of_lives = 6

game_over = False
correct_guesses = []
while not game_over:
    guess = input("Guess a letter: ").lower()

    # 3. Check if the letter the user guessed (guess) is one of the letters in the chosen_word.
    # if the letter is right replace the dash with the letter in the correct position. 
    # If the letter is wrong, print _
    display = ""
    # change the below loop to keep the previously guessed letters in the display variable
    for letter in chosen_word:
        if guess == letter:
            # Replace the dash with the correct letter
            display += letter
            correct_guesses.append(letter)
        elif letter in correct_guesses:
            # Keep the previously guessed letters in the display variable
            display += letter
        else:
            # Keep the dash if the letter is not correct
            display += "-"
    if guess in chosen_word:
        print(f"Good guess! The letter '{guess}' is in the word.")
    else:
        no_of_lives -= 1
        print(f"Wrong guess! You have {no_of_lives} lives remaining.")
    print(display)
    print(f"Lives remaining: {no_of_lives}")

    if display == chosen_word:
        game_over = True
        print("Congratulations! You've guessed the word correctly!")
    elif no_of_lives == 0:
        game_over = True
        print(f"You've run out of lives. The word was '{chosen_word}'. Better luck next time!")
```
