# Dictionaries and Nested List and dictionaries

```python
# {key: value} --> This is a dictionary in Python, 
# where "key" is the identifier used to access the corresponding "value".

programming_dictionaries = {"function": "A piece of code that you can easily call over and over again.", 
"loop": "The action of doing something over and over again.", "list": "A collection of items in a particular order.",
"Bug" : "An error in a program that prevents the program from running as expected."}

print(programming_dictionaries)
```

```python
print(programming_dictionaries)
print(programming_dictionaries.keys())
print(programming_dictionaries.values())
print(programming_dictionaries["loop"])
```

```python
print(programming_dictionaries["Bog"])
```

```python
# Adding new items to the dictionary
programming_dictionaries["bog"] = "A moth that eats clothes."
print(programming_dictionaries["bog"])
```

```python
# removing items from the dictionary
del programming_dictionaries["bog"] 
print(programming_dictionaries["bog"])
```

```python
# wipe out an entire dictionary
programming_dictionaries = {}
print(programming_dictionaries)
```

```python
programming_dictionaries = {"function": "A piece of code that you can easily call over and over again.", 
"loop": "The action of doing something over and over again.", "list": "A collection of items in a particular order.",
"Bug" : "An error in a program that prevents the program from running as expected."}
```

```python
# Loop through the dictionary

for thing in programming_dictionaries:
    print(thing) # This will print the keys of dictionary
    print(programming_dictionaries[thing]) # this will print the values of the dictionary
```

```python
for key in programming_dictionaries.keys():
    print(key) # This will print the keys of dictionary
```

```python
for value in programming_dictionaries.values():
    print(value) # This will print the values of the dictionary
```

```python
for k, v in programming_dictionaries.items():
    print(f"Key: {k}, Value: {v}")
```

```python
# For example
student_scores = {
    'Harry': 88,
    'Ron': 78,
    'Hermione': 95,
    'Draco': 75,
    'Neville': 60
}

student_grades = {}
for k,v in student_scores.items():
    if v >= 91:
        student_grades[k] = 'Outstanding'
    elif v>= 81:
        student_grades[k] = 'Exceeds Expectations'
    elif v>= 71:
        student_grades[k] = 'Acceptable'
    else:
        student_grades[k] = 'Fail'
    
print(student_grades)
```

# Nesting

```python
# {
#     key : [List],
#     key2 : {Nested Dictionary},
# }

# nested List in dictionary
travel_log = {
    "France": ["Paris", "Lille", "Dijon"],
    "Germany": ["Berlin", "Hamburg", "Munich"]
}

# print Lille from the travel_log dictionary
print(travel_log["France"][1])
```

```python
# nested list
nested_list = ["list1", "list2", ["nested_list1", "nested_list2"]]
# print nested_list2 from the nested_list
print(nested_list[2][1])
```

```python
# nested dictionary and lis in a dictionary
travel_log = {
    "France" : {"cities_visited": ["Paris", "Lille", "Dijon"], "total_visits": 12},
    "Germany" : ["Stuttgaet", "Hamburg", "Berlin"]
}

print(travel_log['France']["cities_visited"][2]) # it should print Dijon
```
